# Pi eval runner

The notebook is only the control plane. The runner executes the agent on the OS named by the suite environment. Model inference can be remote (hosted API or a GPU server with an OpenAI-compatible endpoint).

```python
!pip -q install requests pandas
from pi_eval import PiEval
import os
```

## Remote runner

Point this at a runner on a machine with the OS you want to test. Keep the token in Kaggle Secrets / Colab secrets, not in the notebook.

```python
client = PiEval(
    runner_url=os.environ['PI_EVAL_RUNNER_URL'],
    token=os.environ.get('PI_EVAL_RUNNER_TOKEN'),
)
client.health()
```

```python
result = client.run(
    suite='example',
    os_name='windows',
    provider='kaggle',
    model='my-model',
    exports=['messages', 'rendered:chatml.jsonl'],
)
result.report
```

```python
result.traces_frame().head()
result.save('./pi-eval-run')
list(result.datasets)
```

## Local laptop

No runner needed: point `repo_path` at the cloned repository. The CLI then runs on that machine.

```python
local = PiEval(repo_path='/path/to/pi-eval')
local.run(suite='example', provider='openai', model='gpt-5', backend='host')
```
